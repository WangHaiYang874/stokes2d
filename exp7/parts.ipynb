```python
import numpy as np
from matplotlib import pyplot as plt
import sys
sys.path.insert(0,'../src/')


from utils import *
from multiply_connected_pipe import *
from abstract_pipe import *
from curve import *

# Set up plotting parameters
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12
```

```python
t_pipe = TShaped(l=5.0, r=0.5, corner_size=1)
t_pipe.build_geometry(required_tol=1e-3)
t_pipe = RealPipe(t_pipe)

fig,ax = plt.subplots()
t_pipe.plot_plyg_bdr(ax, color='k')
ax.set_aspect('equal')
ax.axis('off')
```

```python
l_pipe = LShaped(l1=5.0, l2=5.0, r=0.5, corner_size=1)
l_pipe.build_geometry(required_tol=1e-3)
l_pipe = RealPipe(l_pipe)

fig,ax = plt.subplots()
l_pipe.plot_plyg_bdr(ax, color='k')
ax.set_aspect('equal')
ax.axis('off')
```

```python
root_of_unity = np.exp(2*np.pi*1j/3)
original = 5j
l1 = [original.real,original.imag]
l2 = [(root_of_unity*original).real,(root_of_unity*original).imag]
l3 = [(root_of_unity**2*original).real,(root_of_unity**2*original).imag]
ls = np.array([l1,l2,l3])

np.arctan2(ls[:,1],ls[:,0])

rs = np.array([0.5,0.5,0.5])

tri_pipe = NLets(np.array([l1,l2,l3]),rs,corner_size=1)
tri_pipe.build_geometry(required_tol=1e-3)
tri_pipe = RealPipe(tri_pipe)

fig,ax = plt.subplots()
tri_pipe.plot_plyg_bdr(ax, color='k')
ax.set_aspect('equal')
ax.axis('off')
```

```python
ls = np.array([[5,0],[0,5],[-5,0],[0,-5]])
rs = np.array([0.5,0.5,0.5,0.5])

cross = NLets(ls,rs,corner_size=1)
cross.build_geometry(required_tol=1e-3)
cross = RealPipe(cross)

fig,ax = plt.subplots()
cross.plot_plyg_bdr(ax, color='k')
ax.set_aspect('equal')
ax.axis('off')
```

```python
root_of_unity = np.exp(2*np.pi*1j/6)
original = 8j
ls = [[(root_of_unity**i*original).real,(root_of_unity**i*original).imag]  for i in range(6)]
ls = np.array(ls)

np.arctan2(ls[:,1],ls[:,0])

rs = np.array([0.5,0.5,0.5,0.5,0.5,0.5])

six_pipe = NLets(ls,rs,corner_size=1)
six_pipe.build_geometry(required_tol=1e-3)
six_pipe = RealPipe(six_pipe)

fig,ax = plt.subplots()
six_pipe.plot_plyg_bdr(ax, color='k')
ax.set_aspect('equal')
ax.axis('off')
```

```python
straight_pipe =    StraightPipe(l=5,r=0.5)
straight_pipe.build_geometry(required_tol=1e-3)
straight_pipe = RealPipe(straight_pipe)

fig,ax = plt.subplots()
straight_pipe.plot_plyg_bdr(ax, color='k')
ax.set_aspect('equal')
ax.axis('off')
```
